```python
import pandas as pd
import numpy as np
```

```python
for_empresas = pd.read_excel(r'C:\Users\jober\Downloads\formacion_empresas_marzo_25.xlsx', sheet_name='Sheet1')
empresas = pd.read_excel(r'C:\Users\jober\Downloads\formacion_empresas_marzo_25.xlsx', sheet_name='Formulario')
```

```python
for_empresas.columns
```

```python
empresas = empresas.merge(for_empresas[['Número de Documento','Fecha de Registro']], on='Número de Documento', how='left')
```

```python
empresas['Fecha de REGISTRO'] = empresas['Fecha de Registro']
empresas = empresas.drop(['Fecha de Registro','HORA'], axis=1)
```

```python
# Get the column you want to move
col_to_move = empresas.columns[6]  # index starts at 0

# Reorder the columns
new_order = [col for col in empresas.columns if col != col_to_move] + [col_to_move]

# Reassign the DataFrame with the new column order
empresas = empresas[new_order]
```

```python
empresas['mes'] = pd.to_datetime(empresas['Fecha de REGISTRO']).dt.month
empresas['año'] = pd.to_datetime(empresas['Fecha de REGISTRO']).dt.year
```
